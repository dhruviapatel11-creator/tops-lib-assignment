```python
import pandas as pd

# Load Excel file
df = pd.read_excel("sales.xlsx")

# Process data in chunks of 20 rows
chunk_size = 20

for start in range(0, len(df), chunk_size):

    chunk = df.iloc[start:start + chunk_size]

    print("Current Chunk:")
    print(chunk)

    print("Number of rows:", len(chunk))
    print("----------------------")
```
