```python
import pandas as pd
import sqlite3
conn = sqlite3.connect("company.db")
emp = pd.read_csv("emp.csv")
emp.to_sql("employees", conn, if_exists="replace", index=False)
print(emp.columns.tolist())
query = "SELECT * FROM employees"
result = pd.read_sql_query(query, conn)
print(result)
conn.close()
```

```python
#4
import pandas as pd
from pathlib import Path

emp_path = Path("emp.csv")
dept_path = Path("dept.csv")

emp = pd.read_csv(emp_path)
dept = pd.read_csv(dept_path)

merged_data = pd.merge(emp, dept, on="Depart_No", how="inner")
print(merged_data[["Name", "Designation", "Department", "City_x", "City_y"]])
print(merged_data)
```
