```python
#1
import pandas as pd
df=pd.read_csv("emp.csv")
duplicates=df[df.duplicated(
    subset=["Name","Depart_No"],
    keep=False
)]
print("duplicated records:")
print(duplicates) 
print("total duplicate records:",len(duplicates))
```

```python
#4
import pandas as pd
df = pd.read_csv("dept.csv")
print("before standardization:")
print(df["Department"])
df["Department"]= (
    df["Department"]
    .astype(str)
    .str.strip()
    .str.title
)
print("after standardization:")
print(df["Department"])
```

```python
#5
import pandas as pd
df = pd.read_csv("titanic.csv")
df["Survived"] = df["Survived"].replace({
    0:"No",
    1:"Yes"
})
print(df[["Name","Survived"]].head(10))
print("Survival status:")
print(df["Survived"].value_counts())
```
