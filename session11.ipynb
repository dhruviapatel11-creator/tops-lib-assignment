```python
import matplotlib.pyplot as plt
import seaborn as sns

iris = sns.load_dataset("iris")
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(iris["sepal_length"], color="pink")
plt.title("Sepal Length")
plt.subplot(1, 2, 2)
plt.plot(iris["petal_length"], color="black")
plt.title("Petal Length")
plt.tight_layout()
plt.show()
```

```python
#2
import matplotlib.pyplot as plt
import seaborn as sns
tips = sns.load_dataset("tips")
day_tips = tips.groupby("day")["tip"].mean()
plt.figure(figsize=(8, 5))
plt.bar(day_tips.index, day_tips.values, color="blue")
plt.xlabel("Day")
plt.ylabel("Average Tip")
plt.title("Average Tip Given on Each Day")

plt.show()
```

```python
#3
import matplotlib.pyplot as plt
import seaborn as sns
iris = sns.load_dataset("iris")
plt.figure(figsize=(8, 5))

plt.plot(iris.index[:50], iris["sepal_length"][:50],
         label="Sepal Length", color="blue")

plt.plot(iris.index[:50], iris["sepal_width"][:50],
         label="Sepal Width", color="red")

plt.xlabel("Index")
plt.ylabel("Measurement")
plt.title("Iris Measurements")
plt.legend()
```

```python
#4
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")

day_tips = tips.groupby("day")["tip"].mean()
plt.Figure(figsize=(8,5))
plt.bar(day_tips.index, day_tips.values)
plt.annotate(
    "highest avg tip",
     xy=(day_tips.values.argmax(), day_tips.max()),
    xytext=(1, day_tips.max() + 0.3),
    arrowprops=dict(facecolor="black", arrowstyle="->")
)

plt.title("Average Tip by Day")
plt.xlabel("Day")
plt.ylabel("Average Tip")
plt.show()
```
