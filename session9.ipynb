```python
#2
import pandas as pd

df = pd.read_excel("sales.xlsx")


month_number = pd.to_datetime(
    df["Month"], format="%B"
).dt.month

df["Date"] = pd.to_datetime({
    "year": df["Year"],
    "month": month_number,
    "day": 1
})

df["Year_Extracted"] = df["Date"].dt.year
df["Month_Extracted"] = df["Date"].dt.month
df["Weekday"] = df["Date"].dt.day_name()
print(df[["Date", "Year_Extracted",
          "Month_Extracted", "Weekday"]].head(10))
```

```python
#3
import pandas as pd
df = pd.read_excel("sales.xlsx")
month_number = pd.to_datetime(
    df["Month"], format="%B"
).dt.month
df["Date"] = pd.to_datetime({
    "year": df["Year"],
    "month": month_number,
    "day": 1
})
df = df.set_index("Date")
monthly_sales = df["Sales"].resample("MS").sum()
print("Monthly Sales:")
print(monthly_sales)
```

```python
#4
import pandas as pd
df = pd.read_excel("sales.xlsx")
df["UTC_Date"] = pd.to_datetime(
    ["2024-01-01 10:00:00"] * len(df),
    utc=True
)
df["IST_date"] = df["UTC_Date"].dt.tz_convert(
    "Asia/Kolkata"
)
print(df[["UTC_Date","IST_date"]].head())
```

```python
#5
import pandas as pd

df = pd.read_excel("sales.xlsx")

month_number = pd.to_datetime(
    df["Month"], format="%B"
).dt.month

df["Date"] = pd.to_datetime({
    "year": df["Year"],
    "month": month_number,
    "day": 1
})
df["is_weekend"] = df["Date"].dt.dayofweek >= 5

print(df[["Date", "is_weekend"]].head(10))
```
