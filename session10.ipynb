```python
#1
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

iris = sns.load_dataset("iris")
plt.figure(figsize=(8,5))
plt.plot(iris.index, iris["sepal_length"], color="blue",
         label="Sepal length")
plt.xlabel("flower index")
plt.ylabel("sepal length")
plt.title("iris dataset")
plt.legend()
plt.grid()
plt.show()
```

```python
#2
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
tips = sns.load_dataset("tips")

day_tips = tips.groupby("day")["tip"].mean()
plt.figure(figsize=(8,5))
plt.bar(day_tips.index, day_tips.values,color="pink")
plt.xlabel("Day")
plt.ylabel("Average Tip")
plt.title("Average Tip Received on Each Day")

plt.show()
```

```python
#3
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

iris = sns.load_dataset("iris")

plt.figure(figsize=(8, 5))

plt.scatter(
    iris["sepal_length"],
    iris["sepal_width"],
    color="green",
    marker="s"
)
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Iris - Sepal Length vs Sepal Width")

plt.show()
```

```python
#4
import pandas as pd
import matplotlib.pyplot as plt
titanic = pd.read_csv("titanic.csv")
age = titanic["Age"].dropna()
plt.Figure(figsize=(8,5))
plt.hist(age, bins=10, color="purple", edgecolor="black")
plt.xlabel("passenger age")
plt.ylabel("no. of passengers")
plt.title("distribution of passenger ages")
plt.show()
```

```python
#5
import pandas as pd
import matplotlib.pyplot as plt
titanic = pd.read_csv("titanic.csv")
survival_count = titanic["Survived"].value_counts().sort_index()
plt.figure(figsize=(7, 7))

plt.pie(
    survival_count.values,
    labels=["Not Survived", "Survived"],
    autopct="%1.1f%%",
)

plt.title("Titanic Passenger Survival Percentage")
plt.show()
```
