```python
#1
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

titanic = pd.read_csv("titanic.csv")
sns.countplot(data=titanic, x="Sex", hue="Survived")
plt.title("Survival Count by Gender")
plt.show()
```

```python
#2
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.barplot(data=tips, x="sex", y="tip", hue="sex")

plt.title("Average Tip by Gender")
plt.show()
```

```python
#3
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset("iris")

sns.boxplot(data=iris, x="species", y="petal_length")

plt.title("Petal Length Comparison")
plt.show()
```

```python
#4
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

titanic = pd.read_csv("titanic.csv")

survival = titanic.groupby("Pclass")["Survived"].mean() * 100

sns.barplot(x=survival.index, y=survival.values)

plt.xlabel("Passenger Class")
plt.ylabel("Survival Percentage")
plt.title("Survival Percentage by Passenger Class")
plt.show()
```

```python
#5
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.histplot(
    data=tips,
    x="total_bill",
    hue="time",
    kde=True
)

plt.title("Lunch and Dinner Bill Distribution")
plt.show()
```
