```python
#1
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset("iris")

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

sns.histplot(data=iris, x="sepal_length", ax=axes[0])
axes[0].set_title("Sepal Length")

sns.scatterplot(
    data=iris,
    x="sepal_length",
    y="petal_length",
    hue="species",
    ax=axes[1]
)
axes[1].set_title("Sepal vs Petal")

sns.boxplot(data=iris, x="species", y="petal_width", ax=axes[2])
axes[2].set_title("Petal Width")

plt.tight_layout()
plt.show()
```

```python
#2
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

titanic = pd.read_csv("titanic.csv")

pivot = titanic.pivot_table(
    values="Survived",
    index="Pclass",
    columns="Sex",
    aggfunc="mean"
)
sns.heatmap(pivot, annot=True, cmap="YlGnBu", fmt=".2f")
plt.title("Survival Rate by Class and Gender")
plt.show()
```

```python
#4
import pandas as pd
import matplotlib.pyplot as plt
titanic = pd.read_csv("titanic.csv")
class_count = titanic["Pclass"].value_counts().sort_index()
plt.figure(figsize=(7, 7))

plt.pie(
    class_count.values,
    labels=["Class 1", "Class 2", "Class 3"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Titanic Passenger Class Distribution")
plt.show()
```
